## Tanscription

In bioinformatins transcription is done through the coding thrain of the DNA, because the opartion is easier (Replace T with U), instad of the biological process of transcibing the copy strain, thich is essentially 2 inversions, plus the t to U replacement.

```python
from Bio.Seq import Seq

sequence = Seq("AGTACACTGGTGC")
print(sequence)
```

```python
#Calculate GC content - GC has a stronger atomic connection
from Bio.SeqUtils import GC123
print(GC1!23(sequence))
```

```python
# the bio process, coding_dna to template_dna to rna
coding_dna = Seq("ATGGCCATTGTAATGGGCCGCTGAAAGGGTGCCCGATAG")
template_dna = coding_dna.reverse_complement()
print(f"Template dna: {template_dna}") #Nuceotide sequences are normally read from the 5' to the 3' direction
#while in the figure the template strand is hown reversed
# now, we need to do the reverse_compelment again, and change "T" to "U"
transcribed_rna = template_dna.reverse_complement_rna()
print(f"Transcribed messenger rna: {transcribed_rna}")
```

```python
#transcribe directly
messenger_rna = coding_dna.transcribe()
messenger_rna
```

```python
# Back transcribuing is obviously possible, too
gen_coding_dna = messenger_rna.back_transcribe()
gen_coding_dna
```

### Translation: RNA-> protein

```python
messenger_rna.translate() # *is a stop colon
```

```python
# We may skip the rna transcription and directly translate from dna (After all, T<->U is the only difference)
coding_dna.translate()
```

## Pairwise Sequence aklignments in Biopython

Pairwise sequences are sequences in different species, which in turn results in similar traits, corresponding to the affected dna sequences

- Global and local allignment is available

When doing alignments,match score and gap pentalties may be specified
1. Match score indicated the compatibility between an alignment of 2 characters in the sequences, Hightly compatible characters should be given positive scores, and incompatible ones should be given negative ones
2. Gap penalties should be negative

Biopython priovides 2 built-in paitwise alligners:
- Bio.pairwise2
- Bio.Align.PairwiseAligner

Focus is on pairwise2, which is decrepated now, and Bio.Align.PairwiseAligner shall be used - Uses configurable class, better documentation, etc. No dealing with the naming conventions of the pairwise2

Global: all matches(with gaps)
Local: best matches only

```python
#Examples of global alignment:
from Bio.Align import PairwiseAligner
#globalxx - matches score 1, mismatsches 0 and no gap penalty
Aligner = PairwiseAligner()
alignments = Aligner.align("ACCGGT", "ACGT")
for alignment in alignments:
    print(alignment)
    print(f"Score: {alignment.score}")
```

## Motif objects in Biopython

A sequence motiv is a nucleotide or amino-acid sequence patern. Sequence motifs aure formed by three-dimensional arrangement of amino acids which may not be adjacent. Biopython provides a seperate module, **Bio.motifs** to acess the functionalities of motif

```python
from Bio import motifs
```

### Creating a simple DNA Motif

Let's create a simple DNA motif sequence

```python
from Bio.Seq import Seq
instances =  [
    Seq("TACAA"),
    Seq("TACGC"),
    Seq("TACAC"),
    Seq("TACCC"),
    Seq("AACCC"),
    Seq("AATGC"),
    Seq("AATGC"), # was duplicated in the orinigal notebook, too
]
```

```python
m = motifs.create(instances)
print(m)
```

```python
print(m.counts)
```

## Plotting Sequence length and GC content

```python
%%time
#Histogram
sizes = [len(rec) for rec in sequence_list]

import matplotlib.pyplot as plt
plt.hist(sizes, bins=40)
plt.title(f"{len(sizes)} items of CAFA6- training data of lengths from {min(sizes)} to {max(sizes)}")
plt.xlabel("Sequence length (bp)")
plt.ylabel("Count")
plt.show()
```

```python
%%time
#Histogram of sequences smaller than 10000
sizes = []
for item in sequence_list:
    if len(item) < 5000:
        sizes.append(len(item))

import matplotlib.pyplot as plt
plt.hist(sizes, bins=40)
plt.title(f"{len(sizes)} items of lengeth-limited CAFA6- training data of lengths from {min(sizes)} to {max(sizes)}")
plt.xlabel("Sequence length (bp)")
plt.ylabel("Count")
plt.show()
```
